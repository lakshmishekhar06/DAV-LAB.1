# K-Means Clustering Using Student Performance Data

## Step 1: Import Required Libraries

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

## Step 2: Create the Student Dataset

```python
# Student data containing study hours and exam scores

X = np.array([
    [1, 25],
    [2, 30],
    [3, 35],
    [4, 40],
    [5, 45],
    [6, 55],
    [7, 60],
    [8, 70],
    [9, 75],
    [10, 80],
    [11, 90],
    [12, 95]
])

print("Student Dataset Shape:", X.shape)
print("\nStudent Data:")
print(X)
```

## Step 3: Standardize the Dataset

```python
# Convert the features into a common scale

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Standardized Student Data:")
print(X_scaled)
```

## Step 4: Find the Optimal Number of Clusters Using Elbow Method

```python
# Calculate inertia for different cluster values

inertia = []

for k in range(1, 11):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_scaled)
    inertia.append(model.inertia_)

# Display the Elbow graph

plt.plot(range(1, 11), inertia, marker='o', color='blue')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Student Data")
plt.grid(True)
plt.show()
```

## Step 5: Apply K-Means Clustering with K = 3

```python
# Divide students into three different clusters

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)

print("Cluster Labels:")
print(clusters)
```

## Step 6: Visualize the Student Clusters

```python
# Plot students according to their assigned clusters

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=clusters,
    cmap='viridis',
    s=100
)

# Display the cluster centers

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker='X',
    s=250,
    color='red',
    label='Centroids'
)

plt.xlabel("Standardized Study Hours")
plt.ylabel("Standardized Exam Scores")
plt.title("Student Performance Clustering (K = 3)")
plt.legend()
plt.grid(True)
plt.show()
```
